```python
%%writefile diabetes_prediction.py

import streamlit as st
import pandas as pd
import numpy as np

df=pd.read_csv("/content/diabetes.csv")

st.title("Diabetes Prediction Application") #Title of the app
st.subheader("Diabetes Data")

#Finding the null values
df.isnull().sum()

#Stats Analysis
df.info()

df.describe()

df.shape

#target_variable->"outcome"
df["Outcome"].value_counts()

#0-> Non-Diabetic
#1-> Diabetic

#grouping the data
df.groupby(["Outcome"]).mean()


#visualising the data
df.groupby(["Outcome"]).mean().plot(kind="bar")

df.pivot_table(index="Outcome",columns="Pregnancies",values="Age",aggfunc="mean").fillna(0)

x=df.drop("Outcome",axis=1)
y=df["Outcome"]

#training and testing
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)

#Shape of the training and testing data
print("x_train shape:",x_train.shape)
print("x_test shape:",x_test.shape)
print("y_train shape:",y_train.shape)
print("y_test shape:",y_test.shape)

from sklearn.linear_model import LogisticRegression
log=LogisticRegression(C=1.0,solver="liblinear",max_iter=100)

log.fit(x_train,y_train)

prediction=log.predict(x_test)
prediction

from sklearn.metrics import accuracy_score
auc_score=accuracy_score(y_test,prediction)

print("accuracy_score:",auc_score)

array=np.array([6,148,72,35,0,33,60,50])
reshaped_array=array.reshape(1,-1)

prediction_array=log.predict(reshaped_array)

def main():
  preg=st.number_input("Diabeties Pregnancies",min_value=0,max_value=20)
  glucose=st.number_input("Glucose:",min_value=0,max_value=200)
  bp=st.number_input("Blood Pressure:",min_value=0,max_value=200)
  skin=st.number_input("Skin Thickness:",min_value=0,max_value=100)
  insulin=st.number_input("Insulin:",min_value=0)
  bmi=st.number_input("BMI:",min_value=0.0)
  dpf=st.number_input("Diabetes Pedigree Function:",min_value=0.0)
  age=st.number_input("Age:",min_value=0,max_value=100)

  diagnosis=" "
  if st.button("diabaties test results:"):
    input_data=[[preg,glucose,bp,skin,insulin,bmi,dpf,age]]
    diagnosis=log.predict(input_data)

    if diagnosis[0]==0:
      st.success("The person is Non-Diabetic")
    else:
      st.error("The person is Diabetic")
main()


```
